OpenAI Python SDK

```python
import os
from openai import OpenAI

client = OpenAI(api_key=os.environ.get("OPENAI_API_KEY"))

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "Bro, tell me a fun fact"
        }
    ],
    model="gpt-5-mini",
)

print(chat_completion.choices[0].message.content)
```

LangChain (langchain-openai)

```python
from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(
    model="gpt-5-mini",
    temperature=0
)

response = llm_openai.invoke("Bro, tell me a fun fact")

print(response.content)
```

Install the required package:

pip install langchain-openai

New LangChain (init_chat_model)

```python
from langchain.chat_models import init_chat_model

llm_openai = init_chat_model(
    model="gpt-5-mini"
)

response = llm_openai.invoke("Hello, how are you?")

print(response.content)
```

Install the required packages:

pip install langchain langchain-openai

note:init_chat_model() is the newer API, but many tutorials still use ChatOpenAI. Both approaches work depending on the LangChain version you're using.
