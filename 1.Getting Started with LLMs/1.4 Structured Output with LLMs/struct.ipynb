```python
from pydantic import BaseModel, Field

class llm_schema(BaseModel):
    setup: str = Field(
        description="The setup for the joke"
    )
    punchline: str = Field(
        description="The punchline for the joke"
    )
```

Structured Output with with_structured_output()

```python
llm_structured_output = llm_openai.with_structured_output(llm_schema)

result = llm_structured_output.invoke("Tell me a joke")

print(result.setup)
print(result.punchline)
```

Using TypedDict

```python
from typing import TypedDict

class llm_schema_td(TypedDict):
    setup: str
    punchline: str
```

```python
obj = llm_schema_td(
    {
        "setup": "setup",
        "punchline": "some punchline"
    }
)
```

Structured Output with TypedDict

```python
llm_structured_typed_dict = llm_openai.with_structured_output(
    llm_schema_td
)

result = llm_structured_typed_dict.invoke("Tell me a joke")

print(result)
```

output:

```python
{
    "setup": "Why did the scarecrow win an award?",
    "punchline": "Because he was outstanding in his field."
}
```
